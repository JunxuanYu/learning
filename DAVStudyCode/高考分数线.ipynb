```python
import pandas as pd

# skiprows=[0] 跳过第0行（大标题行），header=0 使用第1行作为列名
df = pd.read_excel("./data/scores.xlsx", engine="openpyxl", skiprows=[0],sheet_name='scores')

# 重命名列
df.columns = ["年份", "一本文科", "一本理科", "二本文科", "二本理科"]

# 年份转为整数类型
df["年份"] = df["年份"].astype(int)

print("===查看数据===")
print(df.head())
print(df.columns)

data1 = df.copy()
cols = ["一本文科","一本理科","二本文科","二本理科"]

# =========任务1 最高、最低、分差 =========
res1 = pd.DataFrame()
res1["最高分"] = data1[cols].max()
res1["最低分"] = data1[cols].min()
res1["相差分数"] = data1[cols].max() - data1[cols].min()
print("\n====任务1 各分数线最高最低分差====")
print(res1)

# =========任务2 2017与2018对比 2018‑2017=========
row_2017 = data1[data1["年份"]==2017].iloc[0]
row_2018 = data1[data1["年份"]==2018].iloc[0]

res2 = pd.DataFrame()
res2["2017"] = row_2017[cols]
res2["2018"] = row_2018[cols]
res2["分数变化(2018‑2017)"] = row_2018[cols] - row_2017[cols]
print("\n====任务2 2017、2018分数线对比====")
print(res2)

# =========任务3 2006‑2018十三年平均分 =========
df_13 = data1[(data1["年份"]>=2006) & (data1["年份"]<=2018)]
avg_res = df_13[cols].mean()
print("\n====任务3 2006‑2018 13年平均分数线====")
print(avg_res)

# 保存结果到excel
with pd.ExcelWriter(".\data\高考分数线结果.xlsx",engine="openpyxl") as writer:
    res1.to_excel(writer,sheet_name="任务1")
    res2.to_excel(writer,sheet_name="任务2")
    avg_res.to_excel(writer,sheet_name="任务3")
```

```python
df = pd.read_csv('./data/scores.csv',skiprows=[0])
print(df.head())
```

```python
# 重命名列
df.columns = ["年份", "一本文科", "一本理科", "二本文科", "二本理科"]
df["年份"] = df["年份"].astype(int)
print(df.head())
print(df.columns)
```

```python
# 1、一本文理科与二本文理科最高的分数线是多少、最低分数线是多少，相差多少？
data = df.copy()
cols = ["一本文科", "一本理科", "二本文科", "二本理科"]
res1 = pd.DataFrame()
res1['最高分'] = data[cols].max()
res1['最低分'] = data[cols].min()
res1['分差'] = data[cols].max() - data[cols].min()
print(res1)
```

```python
# 2、2017年与2018年相比，一本文理科与二本文理科变化了多少分？
row_2017 = data[data["年份"]==2017].iloc[0]
row_2018 = data[data["年份"]==2018].iloc[0]

res2 = pd.DataFrame()
res2["2017"] = row_2017[cols]
res2["2018"] = row_2018[cols]
res2["分数变化(2018‑2017)"] = row_2018[cols] - row_2017[cols]
print(res2)
```

```python
# 3、求2006--2018年近13年每科分数线的平均值。
df_13 = data[(data["年份"]>=2006) & (data["年份"]<=2018)]
avg_res = df_13[cols].mean()
avg_res = avg_res.round(2)
print(avg_res)
```
