## 3. 查看前几行

`df.head()` 默认返回前 5 行,用于快速确认字段名与数据形态;
`df.tail()` 看最后几行,`df.shape` 看行列数。

```python
df.head()  # 默认查看前 5 行
```

## 4. 统计每列缺失值

`df.isnull()` 返回布尔表,再 `.sum()` 按列求和即得**每列缺失数量**。
这是数据清洗的第一步:先定位缺失,再决定填充或删除。

```python
df.isnull().sum()  # isnull 得到布尔表,sum 按列统计缺失数量
```

## 5. 查看所有列名

`df.columns.tolist()` 把列索引转成普通 Python 列表,便于遍历、筛选或重命名列。

```python
print(df.columns.tolist())  # 把列索引转成普通 list,便于遍历和筛选
```

## 6. 删除缺失过多的列

`deck`(甲板号)缺失率极高且信息量低,直接 `drop(..., axis=1)` 删除。

注意结果赋给了 `df1`,`df` 本身保持不变。

```python
df1 = df.drop('deck',axis=1)  # axis=1 按列删除,结果赋给新变量 df1,原 df 不变
df1.head()  # 查看删除后的表头
```

## 7. 查看各列数据类型

`df.dtypes` 显示每列类型:数值型为 `float64` / `int64`,
类别与文本型为 `object`。数据类型决定了缺失值该怎么填、能不能做数值运算。

```python
df.dtypes  # 查看每列类型,决定缺失值怎么填、能否做数值运算
```

## 8. 统计存活人数

`value_counts()` 按取值出现频次降序排列。这里 `survived` 中 0 = 遇难、1 = 存活。

```python
df1.survived.value_counts()  # 按取值频次降序统计:0 是遇难,1 是存活
```

## 9. 统计舱位等级人数

`pclass` 为舱位等级:1 = 一等座,2 = 二等座,3 = 三等座。

```python
df1.pclass.value_counts()  # 舱位等级人数:1 一等座,2 二等座,3 三等座
```

## 10. 填充缺失的年龄

`age` 列有大量缺失,用 `fillna(20)` 填充。
选 20 是因为泰坦尼克号的船票标题通常按年龄推断,乘客多为二十多岁。

填充后再看一次缺失统计,确认这一列的缺失已经清零。
注意 `df1['age'] = ...` 是在 `df1` 上赋值,原 `df` 不受影响。

```python
df1['age']=df1['age'].fillna(20)  # 用 20 填充年龄缺失,乘客多为二十多岁
df1.isnull().sum()  # 复查:age 列的缺失已清零
```

## 11. 删除缺失的登船地点

`notnull()` 生成布尔掩码,用 `df1[mask]` 筛选出**登船地点非空**的所有行。
这是整行过滤,会同时带走其他列的数据,所以结果要赋给新的 `df2`。

```python
df2 = df1[df1['embarked'].notnull()]  # 布尔掩码筛选出登船地点非空的行,结果赋给新变量
df2.isnull().sum()  # 检查筛选后还剩哪些列有缺失
```

## 12. 交叉表:存活 x 性别

`pivot_table(index='survived', columns='sex', aggfunc='size')`
生成存活与性别的计数交叉表,可以直接对比男女的存活人数。

```python
df2.pivot_table(index='survived',columns='sex',aggfunc='size')  # index 为行、columns 为列,size 统计计数
```

## 13. 相关系数

`corr()` 默认计算皮尔逊相关系数,取值范围 `[-1, 1]`,越接近 ±1 线性关系越强。

此处结果为负,说明舱位等级数值越大(三等座)存活率越低。
注意相关系数只反映线性相关,不代表因果关系。

```python
df['survived'].corr(df['pclass'])  # 皮尔逊相关系数,取值 -1 到 1;结果为负说明舱位等级越低存活率越低
```
