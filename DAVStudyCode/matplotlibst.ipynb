```python
import matplotlib.pyplot as plt
import numpy as np
y = np.arange(1,5)
print(y)
plt.plot(y, c='g',marker='o', label='线1')
plt.plot(y+1,c='0.5',marker='D', label='线2')
plt.plot(y+2,marker='^', label='线3')
plt.plot(y+3,'p', label='线4')
plt.legend()  # 自动读取每条plot里的label
plt.show()
```
